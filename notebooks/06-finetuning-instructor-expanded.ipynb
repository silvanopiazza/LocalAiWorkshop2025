## Tokenization example (what/why/how)

Tokenization converts text into integer token ids. We'll show how many tokens a prompt produces and why that matters for cost and context windows.

```python
# Tokenization demo using a small tokenizer (no model download required for tokenizer)
from transformers import AutoTokenizer
tok = AutoTokenizer.from_pretrained('distilgpt2')
prompt = 'Explain BRCA1 in one sentence.'
tokens = tok(prompt)
print('Tokens:', tokens)
print('Token count:', len(tokens['input_ids']))
```

## Prepare a tiny JSONL dataset (runnable)

We'll create a tiny JSONL file in `labs/data/training_small.jsonl` that matches instruction-following formats.

```python
import json
examples = [
    {
        'instruction': 'Summarize BRCA1 in one sentence.',
        'input': '',
        'output': 'BRCA1 is a gene involved in DNA repair; pathogenic variants increase cancer risk.'
    }
]
p = '../labs/data/training_small.jsonl'
with open(p, 'a') as f:
    for ex in examples:
        f.write(json.dumps(ex) + '
')
print('Appended', len(examples), 'examples to', p)
```
