# 01 — Setup & Environment (Workshop)

This notebook converts the `Course/01-setup-env.md` instructions into an executable, step-by-step setup guide you can run locally. It covers: what to install, recommended environment creation (Conda/Mamba), a pip alternative, optional Docker instructions, GPU notes, and quick verification commands you can run to confirm your environment is ready for the workshop.

## Overview — goals (what you'll end up with)

- A reproducible Python environment for the workshop named `workshop2025` (project-local prefix at `./envs/workshop2025`).
- `pip` / Python packages from `requirements.txt` installed into that environment.
- Optional Docker instructions if you prefer containerized runs.
- A quick smoke test and verification workflow to confirm the setup works on your machine.

## Prerequisites (quick)

- Linux (recommended) or macOS. Windows users: use WSL2 for the smoothest experience.
- 10–20 GB free disk space for Python packages and small model caches (more for large models).
- Optional: NVIDIA GPU + drivers + CUDA if you plan to run models on GPU.
- `git`, `curl`/`wget`, and a terminal (bash).

## Install a Conda-compatible manager (recommended): Mambaforge (fast)

Mambaforge is recommended for speed and reproducibility. If you already have `conda` you can use it instead. The commands below install Mambaforge (Linux example).

```python
# Download & install Mambaforge (Linux x86_64 example).
# If you already have conda/mamba skip these steps.
curl -fsSL https://github.com/conda-forge/miniforge/releases/latest/download/Mambaforge-Linux-x86_64.sh -o Mambaforge.sh
bash Mambaforge.sh -b -p $HOME/mambaforge
# Activate (add to your shell rc file for convenience)
source $HOME/mambaforge/bin/activate
mamba --version || conda --version
```

## Create the project-local Conda env (recommended)

The workshop repo includes `environment.yml` and `requirements.txt`. The recommended, reproducible approach is to create a project-local env under `./envs/workshop2025` so participants can keep environments per-project.
Two options follow: (A) use `mamba/conda` with `environment.yml` and (B) use `python -m venv` + pip (lighter-weight).

```python
# Option A: Create conda env (mamba preferred) at repo-local prefix
# Run this from the repo root (where environment.yml lives)
mkdir -p ./envs
# with mamba (faster)
mamba create -y -p ./envs/workshop2025 python=3.10 -c conda-forge || conda create -y -p ./envs/workshop2025 python=3.10 -c conda-forge
# Install pip packages inside the env using the project's requirements.txt
./envs/workshop2025/bin/python -m pip install --upgrade pip
./envs/workshop2025/bin/pip install -r requirements.txt
# Activate by absolute path in your shell (recommended to avoid naming collisions)
# conda activate ./envs/workshop2025
echo 'To activate:'
echo '  conda activate $(pwd)/envs/workshop2025'
```

```python
# Option B: Lightweight venv + pip (no conda required)
python3 -m venv ./envs/workshop2025-venv
source ./envs/workshop2025-venv/bin/activate
pip install --upgrade pip
pip install -r requirements.txt
echo 'Activated venv at ./envs/workshop2025-venv'
```

## Optional: Docker (containerized) approach

If participants can't install packages locally or you want identical environments across hosts, provide a Docker image. The repository does not currently include a full Dockerfile; for workshops you can create one based on `python:3.10-slim` and install `requirements.txt` — remember to mount the repo and HF cache if you plan to download models.

```python
# Example: build a quick container (optional)
# Place this Dockerfile in repo root and run `docker build -t workshop25 .`
cat > Dockerfile <<'DOCKER'
FROM python:3.10-slim
WORKDIR /workspace
COPY requirements.txt ./
RUN pip install --upgrade pip && pip install -r requirements.txt
CMD ["/bin/bash"]
DOCKER
# Then (once docker is installed):
# docker build -t workshop25 .
# docker run --rm -it -v $(pwd):/workspace workshop25
```
