## A — Recommended: Create the project-local Conda env (one-shot)

Run the following commands from the repository root. They create a local environment at `./envs/workshop2025`, install pip, then install the workshop packages from `requirements.txt`. This is the simplest, most reproducible path for the workshop.

```python
# Make sure you're in the repo root (one line)
cd /home/silvanopiazza/Desktop/workshop25 || exit 1
# Create envs folder
mkdir -p ./envs
# Create the environment (uses mamba if available, falls back to conda)
if command -v mamba >/dev/null 2>&1; then
  mamba create -y -p ./envs/workshop2025 python=3.10 -c conda-forge
else
  conda create -y -p ./envs/workshop2025 python=3.10 -c conda-forge
fi
# Upgrade pip and install requirements into the env
./envs/workshop2025/bin/python -m pip install --upgrade pip
./envs/workshop2025/bin/pip install -r requirements.txt
echo
echo 'Activation (use this in your shell):'
echo '  conda activate '$(pwd)'/envs/workshop2025'
```

## B — Alternative: Lightweight venv (if you cannot use conda)

If you can't use conda/mamba, use the venv flow below. This creates `./envs/workshop2025-venv` and installs the same pip requirements.

```python
# Create and activate a venv (bash)
python3 -m venv ./envs/workshop2025-venv
source ./envs/workshop2025-venv/bin/activate
pip install --upgrade pip
pip install -r requirements.txt
echo 'Activated venv at ./envs/workshop2025-venv'
```

## Troubleshooting (common failures)

- Pip install failed with compilation errors: try the conda/mamba option (conda-forge gives many prebuilt wheels).
- Permission errors during pip: ensure you're using the env's pip (use absolute path).
- `nvidia-smi` missing: you either don't have an NVIDIA GPU or the driver isn't installed — that is OK for the workshop, GPU is optional.
- If a model download fails (network), re-run the smoke test later or ask an instructor to provide a cached model.
