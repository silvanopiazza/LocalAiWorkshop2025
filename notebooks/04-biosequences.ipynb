# Notebook 4 — Working with Biological Sequences (Quickstart)

Runnable examples using Biopython for basic sequence handling and analysis.

```python
# Cell 2: Basic Biopython operations (reverse complement, GC)
try:
    from Bio.Seq import Seq
    from Bio import SeqIO
except Exception as e:
    print('Biopython not installed. Install with: mamba install -c conda-forge biopython')

# Example sequence
s = Seq('ACGTACGTACGT')
print('Sequence:', s)
print('Reverse complement:', s.reverse_complement())
print('GC content (%):', (s.count('G') + s.count('C')) / len(s) * 100)

# Example: create a small FASTA and parse it back
import tempfile
from Bio.SeqRecord import SeqRecord
from Bio.Seq import Seq
records = [SeqRecord(Seq('ACGTACGT'), id='seq1', description='example'), SeqRecord(Seq('TGCATGCA'), id='seq2', description='example2')]
tmp = tempfile.NamedTemporaryFile(mode='w+', suffix='.fasta', delete=False)
SeqIO.write(records, tmp.name, 'fasta')
tmp.flush()
print('
Wrote temporary FASTA to', tmp.name)
for rec in SeqIO.parse(tmp.name, 'fasta'):
    print(f'Read {rec.id}: {str(rec.seq)}')
```
