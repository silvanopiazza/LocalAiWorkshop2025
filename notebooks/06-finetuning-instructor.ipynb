## 1) Why fine-tune? (what/why)

Fine-tuning tailors a pre-trained model to a specific domain or task. LoRA (Low-Rank Adaptation) reduces memory and compute by updating small adapter matrices instead of all model weights — enabling fine-tuning on smaller GPUs.

```python
# Create a tiny JSONL training file (instruction-following format)
import json
examples = [
    {
        'instruction': 'What is the function of TP53?',
        'input': '',
        'output': 'TP53 is a tumor suppressor gene involved in cell cycle regulation and apoptosis.'
    },
    {
        'instruction': 'Summarize: CRISPR enables targeted edits to DNA.',
        'input': '',
        'output': 'CRISPR uses guide RNA and Cas proteins to target and edit DNA sequences.'
    }
]
with open('../labs/data/training_small.jsonl', 'w') as f:
    for ex in examples:
        f.write(json.dumps(ex) + '
')
print('Wrote', len(examples), 'examples to labs/data/training_small.jsonl')
```
