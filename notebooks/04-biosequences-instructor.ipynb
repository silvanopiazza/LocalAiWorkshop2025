## 1) Reading FASTA (what/why)

What: Parse FASTA files with Biopython.
Why: FASTA is a common format for storing sequences; parsing is the first step in any pipeline.

```python
# Example: read the sample FASTA from labs/data (created in repo)
from pathlib import Path
fasta = Path('../labs/data/sample_fasta.fasta')
try:
    from Bio import SeqIO
    for rec in SeqIO.parse(str(fasta), 'fasta'):
        print(rec.id, len(rec.seq), str(rec.seq)[:60])
except Exception as e:
    print('Biopython not installed or sample file missing:', e)
```

## 2) Basic sequence operations (what/why)

Operations: reverse complement, GC content, translation. Why: Common preprocessing and QC steps in many analyses.

```python
# Basic sequence ops
try:
    from Bio.Seq import Seq
    s = Seq('ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG')
    print('Sequence:', s)
    print('Reverse complement:', s.reverse_complement())
    gc = (s.count('G') + s.count('C')) / len(s) * 100
    print(f'GC content: {gc:.1f}%')
    print('Translated (first 20 aa):', s.translate()[:20])
except Exception as e:
    print('Biopython not available:', e)
```
