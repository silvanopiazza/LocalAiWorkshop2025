# Notebook 6 — Fine-tuning & LoRA (Guided Quickstart)

This notebook provides safe, small-scale examples for dataset formatting and a non-training demonstration of preparing data for LoRA fine-tuning. It does NOT run large training by default (keeps things CPU-friendly).

```python
# Cell 2: Create a tiny JSONL training file (instruction-following format)
import json
examples = [
    {
        'instruction': 'What is the function of TP53?',
        'input': '',
        'output': 'TP53 is a tumor suppressor gene involved in cell cycle regulation and apoptosis.'
    },
    {
        'instruction': 'Summarize this abstract: CRISPR enables targeted edits.',
        'input': '',
        'output': 'CRISPR is a gene-editing system that uses guide RNA and Cas proteins to target DNA sequences.'
    }
]
with open('training_small.jsonl', 'w') as f:
    for ex in examples:
        f.write(json.dumps(ex) + '
')
print('Wrote training_small.jsonl with', len(examples), 'examples')
```
