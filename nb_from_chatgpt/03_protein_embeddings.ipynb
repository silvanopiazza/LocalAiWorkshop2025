# Protein Sequence Embeddings (k‑mer baseline)

**Goal:** create simple protein embeddings using k‑mer frequencies (works offline).

## Why k‑mers?
K‑mer counts approximate local sequence patterns. They are a good lightweight baseline when protein LMs (ESM/ProtT5) aren't available.

```python
from collections import Counter
import itertools
import numpy as np
import pandas as pd

seqs = {
    'P1': 'MEEPQSDPSV',   # toy p53 fragment
    'P2': 'MSSSEEVSWG',
    'P3': 'MDLSALRVEE',
}

k = 2
alphabet = list('ACDEFGHIKLMNPQRSTVWY')
k_mers = [''.join(p) for p in itertools.product(alphabet, repeat=k)]

 def kmer_vec(seq, k, vocab):
    km = [seq[i:i+k] for i in range(len(seq)-k+1)]
    cnt = Counter(km)
    return np.array([cnt.get(t,0) for t in vocab], dtype=float)

emb = {name: kmer_vec(seq, k, k_mers) for name, seq in seqs.items()}
M = np.vstack([emb[k] for k in seqs.keys()])

df = pd.DataFrame(M, index=list(seqs.keys()), columns=k_mers)
print('Shape:', df.shape)
print(df.iloc[:, :12].head())
```
