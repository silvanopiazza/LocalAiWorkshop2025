# Bio Applications: Summaries & DNA k‑mers

**Part A: Summarize a short abstract (rule‑based)**
When an LLM is unavailable, a simple sentence scorer can still produce a TL;DR.

**Part B: DNA k‑mer encoding**
Build a 3‑mer vector for DNA strings (ACGT).

```
# Part A: crude extractive summary
import re
text = (
    'Breast cancer involves genomic instability. DNA repair pathways like BRCA1/2 are key. '
    'We analyze RNA‑seq profiles to identify dysregulated genes and pathways. '
    'Results highlight homologous recombination and cell cycle checkpoints as top hits.'
)
sents = re.split(r'(?<=[.!?])\s+', text)
print('Sentences:', sents)
# naive: keep first and longest
summary = ' '.join(sorted(sents, key=len, reverse=True)[:1])
print('\nTL;DR:', summary)

# Part B: DNA 3‑mer encoding
from collections import Counter
import itertools
import numpy as np
alphabet = 'ACGT'
kmers = [''.join(p) for p in itertools.product(alphabet, repeat=3)]

def encode_3mer(seq):
    seq = seq.upper().replace('N','')
    toks = [seq[i:i+3] for i in range(len(seq)-2)]
    c = Counter(toks)
    return np.array([c.get(k,0) for k in kmers], dtype=float)

seq = 'ACGTTGACCTGA'
vec = encode_3mer(seq)
print('Vector length:', len(vec))
print('Nonzero entries:', int(np.count_nonzero(vec)))
```
