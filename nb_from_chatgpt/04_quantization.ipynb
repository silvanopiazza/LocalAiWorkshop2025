# Quantization Demo (NumPy)

**Goal:** show how quantization compresses weights/vectors and introduces a small error.

## Steps
- Create a float32 matrix
- Quantize to int8 with a scale factor
- Reconstruct and compute mean absolute error (MAE)

```
import numpy as np
np.random.seed(0)
W = np.random.randn(1000, 256).astype('float32')
scale = np.max(np.abs(W)) / 127.0
W_q = np.clip(np.round(W/scale), -128, 127).astype('int8')
W_rec = (W_q.astype('float32') * scale)
mae = np.mean(np.abs(W - W_rec))
print('Original bytes:', W.nbytes)
print('Quantized bytes:', W_q.nbytes)
print('Compression ratio: {:.2f}x'.format(W.nbytes / W_q.nbytes))
print('MAE:', mae)
```
