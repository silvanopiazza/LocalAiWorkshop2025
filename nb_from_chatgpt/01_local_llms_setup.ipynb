## 2. Example: call a local model via HTTP
If you have **Ollama** running locally, this POST request will stream a response for the `llama3` model. Replace the model name if you have a different one.

## 3. Minimal prompt engineering
A simple structure that works well for factual Q&A and code: *system → instruction → context → constraints*.

```python
import sys, platform
print(platform.python_version())
print(sys.executable)
```

```python
import json, urllib.request
req = urllib.request.Request(
    url='http://localhost:11434/api/generate',
    data=json.dumps({
        'model': 'llama3',
        'prompt': 'List 3 benefits of running LLMs locally for bioinformatics.',
        'stream': False
    }).encode('utf-8'),
    headers={'Content-Type': 'application/json'}
)
try:
    with urllib.request.urlopen(req, timeout=5) as r:
        out = json.loads(r.read().decode('utf-8'))
        print(out.get('response','<no response>'))
except Exception as e:
    print('Could not reach a local model (is Ollama running?):', e)
```

```python
text = (
    'You are a helpful assistant for biology researchers.\n'
    'Instruction: Summarize what a transformer does in 2 bullets.\n'
    'Constraints: Be precise and concise.'
)
print(text)
```
