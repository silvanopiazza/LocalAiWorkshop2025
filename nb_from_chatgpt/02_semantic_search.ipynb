# Building a Tiny Semantic Search Engine (TF‑IDF + Cosine)

**Goal:** index a small corpus and retrieve the most similar documents to a query.

## Steps
1. Clean text → 2. Vectorize with `TfidfVectorizer` → 3. Compute cosine similarity → 4. Rank and inspect matches.

```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

corpus = [
    'BRCA1 regulates DNA repair in breast cancer',
    'miR-30d downregulates targets involved in EMT',
    'p53 is a tumor suppressor gene involved in cell cycle',
    'Extracellular vesicles carry miRNA cargo in TNBC',
]

vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(corpus)

query = 'DNA repair in breast tumors'
q_vec = vectorizer.transform([query])
scores = cosine_similarity(q_vec, X).ravel()

ranked = pd.DataFrame({
    'doc': corpus,
    'score': scores
}).sort_values('score', ascending=False)
print('Query:', query)
print(ranked)
```
